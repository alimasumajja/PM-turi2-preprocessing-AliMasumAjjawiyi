```python
import pandas as pd
import numpy as np

data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
df
```

```python
df['usia'] = df['usia'].fillna(df['usia'].mean())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
print(df['usia'])
print(df['pendapatan'])
```

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
print(X)
print(y)
```

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2}) # ordinal
X['status'] = ['menikah', 'belum menikah', 'menikah', 'belum menikah', 'belum menikah', 'menikah', 'belum menikah', 'menikah', 'belum menikah', 'belum menikah']
X = pd.get_dummies(X, columns=['kota', 'status'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})

# print(X['pendidikan'])
print(X.columns)

print(X)

print(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
X, y, test_size=0.3, random_state=42)
print('Data Latih: ', len(X_train))
print('Data Uji: ', len(X_test))
```

```python
from sklearn.preprocessing import StandardScaler
num = ['usia','pendapatan','pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num]) # cegah data leakage
print(X_train[num].describe())
print(X_train[num])
print(X_test[num])
```

```python
# 1. Verifikasi Tidak Ada Nilai Hilang
# Hasil yang diharapkan: Semua kolom bernilai 0
print("--- Pengecekan Nilai Hilang ---")
print(X_train.isnull().sum())

# 2. Verifikasi Semua Data Numerik
# Hasil yang diharapkan: Semua tipe data adalah int atau float, tidak ada 'object'
print("\n--- Pengecekan Tipe Data ---")
print(X_train.dtypes)

# 3. Verifikasi Skala Seragam (Hasil StandardScaler)
# Hasil yang diharapkan pada kolom 'usia', 'pendapatan', 'pendidikan':
# Nilai rata-rata (mean) mendekati 0 dan standar deviasi (std) mendekati 1
print("\n--- Pengecekan Skala Data ---")
print(X_train.describe().round(2))
```
